### Setup

```python
import numpy as np
from sklearn.tree import DecisionTreeRegressor
```

```python
np.random.seed(0)
n = 40
X = np.sort(np.random.uniform(-4, 4, n))
y = np.sin(X) * 3 + np.random.normal(0, 0.4, n)
X_col = X.reshape(-1, 1)
```

```python
F0 = np.mean(y)
F = np.full(n, F0)
lr = 0.3
M = 50
mse_history = []
snap = {}
```

```python
for m in range(1, M+1):
    r = y - F
    tree = DecisionTreeRegressor(max_depth=2)
    tree.fit(X_col, r)
    F = F + lr * tree.predict(X_col)
    mse_history.append(np.mean((y - F) ** 2))
    if m in [1, 10, 20, 50]:
        snap[m] = F.copy()
```

```python
print("MSE at F0:", round(np.mean((y - F0) ** 2), 3))
for m in (1, 5, 20, 50):
    print(f"MSE after {m:>2} trees:", round(mse_history[m - 1], 4))
```
